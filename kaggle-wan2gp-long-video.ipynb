# فيديو طويل مجاناً — Wan2GP على Kaggle GPU

دفتر جاهز يفتح واجهة **Wan2GP** (مستودع `deepbeepmeep/Wan2GP`، فُحص حياً 2026-11-23) داخل Kaggle مجاناً، لتوليد فيديو طويل بالربط المتسلسل — بدل مقاطع الثواني المعدودة في Hugging Face ZeroGPU.

## قبل التشغيل — فعّل في Kaggle
- **Settings → Accelerator: GPU T4 x2**، و**Internet: ON** (بدون إنترنت لن يثبّت شيئاً).
- تحتاج **حساب Kaggle + تحقق برقم الهاتف** لتفعيل الـGPU — هذا الطريق «بلا دفع» لا «بلا حساب».

## الحصة بصراحة (نوفمبر 2026)
- نحو **30 ساعة GPU أسبوعياً**، تتجدد **السبت 00:00 UTC**؛ الجلسة حتى **12 ساعة**، وتنقطع بعد **20 دقيقة** خمول.
- منذ 2026-09-15 بطاقة Kaggle المجانية هي **T4 x2 (‏2×16GB)**؛ ساعة على T4x2 = ساعة من الحصة.
- مخرجات محفوظة حتى 20GB — **نزّل الفيديو قبل إنهاء الجلسة**، فالرابط العام يموت بموتها.

## ماذا يولّد؟
- **Wan 2.1 / 2.2**: بلا صوت أصلاً — صوت لاحق عبر MMAudio داخل الواجهة.

## إعدادات مقترحة للفيديو الطويل على T4 المجانية

- **النموذج**: LTX-2 المقطّر (distilled) إن أردت صوتاً؛ أو Wan 2.1 1.3B إن أردت الأخف.
- **الدقة**: 512×384، و**81 إطاراً** للمقطع الواحد (~3.4 ثوانٍ).
- **الربط**: من آخر إطار في كل مقطع ابدأ المقطع التالي (continue video / Showrunner) حتى تصل إلى الدقائق.
- **الحفظ**: نزّل كل مقطع MP4 من الواجهة قبل إغلاق الجلسة، ثم ادمج المقاطع بأي محرر فيديو (أو تبويب «فيديو طويل» في الاستوديو).
